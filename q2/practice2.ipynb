# 인사이트
- 컨테이너 물동량이 감소한 연도: 2016, 2020, 2022
- 환적증감율이 감소한 연도: 2016, 2022
- 2020년 컨테이너 물동량이 감소했는데, 환적증감율은 증가했다. 왜 그럴까?

- 2014년부터 환적 비중이 전체 물동량에서 절반 이상으로 나타났다.
